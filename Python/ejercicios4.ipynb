# Fizzbuzz
- Múltiplos de 3 por la palabra "fizz".
- Múltiplos de 5 por la palabra "buzz".
- Múltiplos de 3 y de 5 a la vez por la palabra "fizzbuzz".

```python
def fizzbuzz():
    for i in range(1,101):
        print(i, end=" | ")
        if (i%3==0) and (i%5==0):
            print("fizzbuzz")
        elif i%3 == 0:
            print("fizz")
        elif i%5 == 0:
            print("buzz")
        
        else:
            print("-")

fizzbuzz()
```

# factorial

```python
def factorial_rec(n: int):
    if n < 2:
        return 1
    return n * factorial_rec(n-1)

def factorial(n: int):
    fact = 1
    for i in range(1, n+1):
        fact *= i
    return fact

for i in range(1,11):
    fact = factorial_rec(i)
    print(f"{i}! = {fact}")
```

# Baseball Game

- You are keeping the scores for a baseball game with strange rules. At the beginning of the game, you start with an empty record.

- You are given a list of strings operations, where operations[i] is the ith operation you must apply to the record and is one of the following:

- An integer x: Record a new score of x.
- '+': Record a new score that is the sum of the previous two scores.
- 'D': Record a new score that is the double of the previous score.
- 'C': Invalidate the previous score, removing it from the record.
- Return the sum of all the scores on the record after applying all the operations.

- The test cases are generated such that the answer and all intermediate calculations fit in a 32-bit integer and that all operations are valid.

```python
def baseball_game(ops: list) -> int:
    record = []
    for score in ops:
        if score == "+":
            record.append(record[-2] + record[-1])
        elif score =="D":
            record.append(2*(record[-1]))
        elif score == "C":
            record.pop()
        else:
            record.append(int(score))
    return sum(record)

ops = ["5","2","C","D","+"]
res = baseball_game(ops)
print(ops, "\noutput = ", res, end="\n\n")

ops = ["5","-2","4","C","D","9","+","+"]
res = baseball_game(ops)
print(ops, "\noutput = ", res, end="\n\n")

ops = ["1","C"]
res = baseball_game(ops)
print(ops, "\noutput = ", res)
```

# Longest substring

```python
def longest(string: str) -> dict:
    length = len(string)
    curr_subs = ""
    long_subs = ""
    chars = {}

    for i in range(length):
        for j in range(i, length):
            if string[j] not in chars:
                chars[string[j]] = string[j]
                curr_subs += string[j]
            else:
                if len(curr_subs) > len(long_subs):
                    long_subs = curr_subs
                break
        curr_subs = ""
        chars = {}
    return {
        "length": len(long_subs),
        "substring": long_subs
    }

def longest_opt(string: str) -> int: # version optimizada
    chars = set()
    left = 0
    max_lengt = 0

    for right in range(len(string)):
        while string[right] in chars: # si hay repetido
            chars.remove(string[left])
            left +=1
        chars.add(string[right])
        max_lengt = max(max_lengt, right-left+1)
    return max_lengt

#### PRUEBAS
string = "abcabcbb"
res = longest(string)
print(res)

string = "bbbbbb"
res = longest(string)
print(res)

string = "abcabcdbb"
res = longest(string)
print(res)

print("#### VERSION 2")
string = "abcabcbb"
res = longest_opt(string)
print(res)

string = "bbbbbb"
res = longest_opt(string)
print(res)

string = "abcabcdbb"
res = longest_opt(string)
print(res)
        
```

# Majority Element

```python
from random import randint

def majority_element(arr: list) -> int:
    elements = {}
    max_elm = [-1,0] # (elemento , conteo)
    mid = len(arr)//2

    for e in arr:
        if e in elements:
            elements[e] += 1
            if elements[e] > mid: # si el elemento lleva un conteo superior a la mitad de la cantidad total de elementos
                return e
        else:
            elements[e] = 1

        max_elm = [e, elements[e]] if max_elm[1] < elements[e] else max_elm

    return max_elm[0]


nums = [2,2,1,1,1,2,2]
res = majority_element(nums)
print(res, end="\n\n")

nums = [randint(1,10) for _ in range(10)]
print(nums)
res = majority_element(nums)
print(res, end="\n\n")

nums = [randint(1,5) for _ in range(10)]
print(nums)
res = majority_element(nums)
print(res, end="\n\n")
    
```

# Palindromo

```python
def palindromo(string: str) -> bool:
    i = 0
    j = len(string)-1
    while (i < j):
        if string[i] != string[j]:
            return False
        i += 1
        j -= 1
    return True


string = "abcdefg"
res = palindromo(string)
print(res)

string = "abba"
res = palindromo(string)
print(res)

string = "abcba"
res = palindromo(string)
print(res)

string = "abbac"
res = palindromo(string)
print(res)
```

# Invertir cadena

```python
def invertir_cadena(text: str) -> str:
    chars = ["" for _ in range(len(text))]
    i = 0
    j = len(text)-1
    while (i < j):
        chars[i], chars[j] = text[j], text[i]
        i+=1
        j-=1
    if i==j: # si la cadena es impar se asigna el elemento central
        chars[i] = text[i]
    return "".join(chars)


text = "abcdefg"
text = invertir_cadena(text)
print(text)

text = "python"
text = invertir_cadena(text)
print(text)
```

# Busqueda binaria

```python
from random import randint

def binary_search(arr: list, target: int) -> int:
    i = 0
    j = len(arr)-1
    while i<=j:
        mid = i + ((j-i)//2)
        if arr[mid] == target:
            return mid
        elif arr[mid] > target:
            j = mid-1
        else:
            i = mid+1
    return None


for _ in range(5):
    elements = [randint(1,50) for _ in range(20)]
    elements.sort()
    target = randint(1,50)
    res = binary_search(elements, target)
    print("elements =", elements)
    print("target =", target)
    print(f"{elements[res] if res!=None else 'No encontrado'}", end="\n\n")
```
